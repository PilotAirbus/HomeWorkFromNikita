```python
import pandas as pd
```

```python
df = pd.read_csv('SWW.csv')
print(df)
```

```python
find_data = df[(df['Спорт'] == 'Футбол') & (df['Оценка'] > 80)]
print(find_data)
```

```python
group_data = df.groupby('Спорт')['Оценка'].mean()
print(group_data)
```

```python
group_data.to_csv('SWW1.csv')
```
